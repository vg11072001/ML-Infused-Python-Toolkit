```python
'''
To build an Adjacency List 
'''

# 1. Graph Node used for Adjaceny List
class GraphNode:
    def __init__(self, val):
        self.val = val
        self.neighbors = []

# 2. use Hashmap : 
adjList = {"A":[], "B":[]}


'''
In Case want to build an Adjacency list following code can be used
'''
# Initialize an empty dictionary
adjList = {}

# Given di"rected edges, to build an Adjacency list
edges = [["A", "B"], ["B", "C"], ["B", "E"], ["C", "E"], ["E", "D"]]

# now to create the adjacency list

# src : is source and
# dst : is destination
for src, dst in edges:
    
    if src not in adjList:
        adjList[src] = []
    
    if dst not in adjList:
        adjList[dst] = []
    
    adjList[src].append(dst)

print("adjList")
print(adjList)

'''
Now Lets see DFS and BFS
'''

# 1. DFS (backtracking) to count paths 
def dfs(node, target, adjList, visit):
    # to not visit same node
    if node in visit:
        return 0
    
    if node == target:
        return 1
    
    count = 0
    
    visit.add(node)
    
    for neighbors in adjList[node]:
        count += dfs(neighbors, target, adjList, visit)
        
    visit.remove(node)
    
    return count

dfsCountPath = dfs("A", "E", adjList, set())
print("dfsCountPath")
print(dfsCountPath)

# Time Complexity:
# O(N^V) where N : average number of edges on each node and V is vertices means nodes
# not much efficient


# 2. BFS

from collections import deque

def bfs(node, target, adjList):
    # to track the once we already visited
    visit = set()
    visit.add(node)

    length = 0
    
    # the queue storing the nodes on neighbors
    queue = deque()
    queue.append(node)
    
    
    while queue:
        for i in range(len(queue)):
            curr = queue.popleft()
            
            if curr == target:
                return length
            
            for neighbors in adjList[curr]:
                if neighbors not in visit:
                    visit.add(neighbors)
                    
                    queue.append(neighbors)
                    
        length += 1
        
    return length

bfsCountPath = bfs("A", "E", adjList)
print("bfsCountPath")
print(bfsCountPath)


# Time Complexity :  O(V + N) 
# Space: O(V)
```
